```python
%%writefile DSA0406_LAB_EXP/numpy_subject_average.py
import numpy as np

subjects = ["Math", "Science", "English", "History"]

student_scores = np.array([
    [85, 78, 92, 88],
    [90, 82, 75, 80],
    [70, 95, 85, 78],
    [88, 76, 90, 92]
])

subject_avg = np.mean(student_scores, axis=0)
best_index = np.argmax(subject_avg)

for name, avg in zip(subjects, subject_avg):
    print(f"{name}: {avg:.2f}")

print(f"\nHighest average subject: {subjects[best_index]} ({subject_avg[best_index]:.2f})")
```
